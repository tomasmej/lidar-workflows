# Creating 3D Interactive Surface Plot From Raster DEMs

```python
from whitebox_workflows import WbEnvironment, download_sample_data, show


wbe = WbEnvironment()
# 3D surface plot example


wbe.working_directory = download_sample_data('mill_brook_dem')

dem = wbe.read_raster('dem.tif')
show(
    dem,
    figsize = (10, 8),
    skip = 2,
    plot_as_surface = True,
    vert_exaggeration = 2.5,
    color = 'lightgreen',
    shade = True,
    linewidth = 0.0,
    rcount = 175, #default is 50 rows and columns for grid 
    ccount = 175,
    antialiased = False
)
```
